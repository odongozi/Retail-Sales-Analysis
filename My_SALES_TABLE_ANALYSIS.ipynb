## 3. Data Merging

The datasets were merged to create a single table containing the information required for the analysis.

The sales table was joined with the sales team, product, and store location tables using their corresponding index or key columns.

```python
table=table.merge(products,
                  left_on="Productindex",
                  right_on="Index",
                  how="left")
```

```python
table =table.merge(store_locations,
                   left_on="Storeindex",
                   right_on="id",
                   how="left")
```

## 5. Data Transformation

Additional columns were created from the existing data to support the analysis.

For example, revenue was calculated using the order quantity and unit price.

```python
#ADDING REVENUE COLUMN
table["revenue"]=table["Order qty"]*table["unit price"]
```

```python
#ADDING PROFIT COLUMN
table["profit"]=(table["unit price"]-table["unit cost"])*table["Order qty"]
```

```python
#ADD MONTHS COLUMN TO MY TABLE
table["month"]=table["Sales Date"].dt.to_period("M")
```

## 7. Product Analysis/Data Visualization

Product performance was analyzed to identify the products generating the highest revenue, profit, and sales quantity.
Charts were created using Matplotlib to provide a visual representation of the major findings from the analysis.

```python
#PRODUCT BY REVENUE
product_revenue=table.groupby("Product Name")["revenue"].sum().sort_values(ascending=False)
product_revenue=product_revenue.head(10)
print(product_revenue.to_string(float_format='{:,.0f}'.format))
```

```python
plt.figure(figsize=(6,6))
plt.bar(product_revenue.index,product_revenue.values)
plt.title("TOP 10 PRODUCT BY REVENUE")
plt.xlabel("REVENUE")
plt.ylabel("PRODUCTS")
plt.xticks(rotation=45)
plt.show()
```

```python
#Revenue by Sales Channel
sales_channel_revenue = table.groupby("Sales Channel")["revenue"].sum().sort_values(ascending=False)
print(sales_channel_revenue.to_string(float_format='{:,.0f}'.format))
```

```python
#QUANTITY SOLD BY PRODUCT
product_quantity=table.groupby("Product Name")["Order qty"].sum().sort_values(ascending=False)
product_quantity=product_quantity.head(5)
product_quantity

```

```python
plt.figure(figsize=(6,3))
plt.barh(product_quantity.index,product_quantity.values)

plt.title("TOP 5 PRODUCT BY QUANTITY SOLD")
plt.xlabel("PRODUCT QUANTITY")
plt.ylabel("PRODUCT")
plt.xticks(rotation=45)

plt.show()
```

```python
#MONTHLY SALES ANALYSIS
monthly_revenue=table.groupby("month")["revenue"].sum().sort_values(ascending=False)
print(monthly_revenue)
```

```python
#TOP 5 PRODUCT BY PROFIT
product_profit=table.groupby("Product Name")["profit"].sum().sort_values(ascending=False)

product_profit=product_profit.head(5)
print(product_profit.to_string(float_format='{:,.0f}'.format))
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(6,6))
plt.bar(product_profit.index,product_profit.values)

plt.title("TOP 5 PRODUCT BY PROFIT")
plt.xlabel("PRODUCT")
plt.ylabel("PROFIT")
plt.xticks(rotation=45)

plt.show()
```

## 8. Sales Team Analysis

Sales team performance was evaluated based on revenue, profit, number of orders, and quantity sold.

```python
#Sales TEAM ANALYSIS(GETTING THE PERFORMANCE)
team_performance = table.groupby("Sales Team").agg(
    Revenue=("revenue", "sum"),
    Profit=("profit", "sum"),
    Orders=("Order Number", "nunique"),
    Quantity=("Order qty", "sum")).sort_values("Profit", ascending=False)
print(team_performance.to_string(float_format='{:,.0f}'.format))
```
